```python
struct_table = struct_catalog[0]
```

```python
print(len(struct_table))           # number of rows (galaxies)
print(struct_table.colnames)       # list of all column names
```

```python
from astropy.coordinates import SkyCoord
import astropy.units as u

mass_table = catalog[0]   # your original Mendel14 table

coords_mass = SkyCoord(ra=mass_table['_RA'], dec=mass_table['_DE'])
coords_struct = SkyCoord(ra=struct_table['_RA'], dec=struct_table['_DE'])

idx, sep2d, _ = coords_mass.match_to_catalog_sky(coords_struct)

# Keep only good matches (within 1 arcsecond, a safe threshold for same-survey coordinates)
good_match = sep2d < 1*u.arcsec
print(f"{good_match.sum()} out of {len(mass_table)} matched")
```

```python
import matplotlib.pyplot as plt
import numpy as np

idx, sep2d, _ = coords_mass.match_to_catalog_sky(coords_struct)

plt.hist(sep2d.arcsec, bins=100, range=(0, 2))
plt.xlabel('Separation (arcsec)')
plt.ylabel('Count')
plt.yscale('log')
plt.show()

print(np.percentile(sep2d.arcsec, [50, 90, 99, 99.9]))
```

```python
values, counts = np.unique(idx[good_match], return_counts=True)
duplicates = values[counts > 1]
print(f"{len(duplicates)} struct_table rows matched more than once")
```
